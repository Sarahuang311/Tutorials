```python
# Part 2
num_hidden_uints = 28

def sigmoid(x):
    y = 1/(1 + np.exp(-x))
    return y
def diff_sigmoid(x):
    y = x*(1-x)
    return y

def nnet(train_x, train_y, test_x, test_y, lr, num_epochs):
    num_train = len(train_y)
    num_test = len(test_y)
    
    train_x = np.hstack((train_x, np.ones(num_train).reshape(-1,1))) # q7_1
    test_x = np.hstack((test_x, np.ones(num_test).reshape(-1,1)))
    
    num_input_uints = train_x.shape[1]  # 785 
    
    wih = np.random.uniform(low=-1, high=1, size=(num_hidden_uints, num_input_uints)) #28*785
    who = np.random.uniform(low=-1, high=1, size=(1, num_hidden_uints+1)) # 1 * 29
    
    w1 = np.zeros([784, 28])
    w2 = np.zeros([1,])

    for epoch in range(1, num_epochs+1):
        out_o = np.zeros(num_train) # q7_2
        out_h = np.zeros((num_train, num_hidden_uints+1))  #num_train * 29 # q7_3
        out_h[:,-1] = 1 #q7_4
        for ind in range(num_train): # q7_5
            row = train_x[ind]  # len = 785 #q7_6
            out_h[ind, :-1] = sigmoid(np.matmul(wih, row)) #q7_7
            out_o[ind] = 1 / (1 + np.exp(-sum(out_h[ind] @ who.T))) #q7_7

            delta = np.multiply(diff_sigmoid(out_h[ind]), (train_y[ind] - out_o[ind]) * np.squeeze(who))
            wih += lr * np.matmul(np.expand_dims(delta[:-1], axis=1), np.expand_dims(row,axis=0))
            who += np.expand_dims(lr * (train_y[ind] - out_o[ind]) * out_h[ind,:], axis=0)
        error = sum(- train_y * np.log(out_o) - (1-train_y) * np.log(1-out_o))
        num_correct = sum((out_o > 0.5).astype(int) == train_y)
        
        print('epoch = ', epoch, ' error = {:.7}'.format(error), 'correctly classified = {:.4%}'.format(num_correct / num_train))
        
    return wih, who
```
